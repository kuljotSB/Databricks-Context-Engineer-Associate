### Creating the OpenAI Client

```python
from openai import OpenAI

client = OpenAI(
    api_key = "YOUR_DATABRICKS_ACCESS_TOKEN",
    base_url = "YOUR_AI_GATEWAY_ENDPOINT"
)
```

### See Traffic Splitting in Action

```python
from collections import Counter

# Store which model handled each request
model_counts = Counter()

# Send 10 identical requests
for i in range(10):

    completion = client.chat.completions.create(
        model="YOUR_COMPLETE_MODEL_NAME",
        messages=[
            {
                "role": "system",
                "content": "You are a helpful AI assistant"
            },
            {
                "role": "user",
                "content": "Hi, I am Kuljot. Nice to meet ya!"
            }
        ]
    )

    # Extract response
    assistant_response = completion.choices[0].message.content

    # Extract the model that actually handled the request
    model_used = completion.model

    # Count requests handled by each model
    model_counts[model_used] += 1

    # Display individual request result
    result = {
        "assistant_response": assistant_response,
        "model": model_used
    }

    print(f"Request {i + 1}:")
    print(result)
    print("-" * 60)


# Display traffic distribution
print("\nTraffic Distribution")
print("=" * 60)

for model, count in model_counts.items():
    print(f"{model}: {count} requests")

print("=" * 60)
print(f"Total requests: {sum(model_counts.values())}")
```
